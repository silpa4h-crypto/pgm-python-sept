```python
#list operations
listA = list(map(int, input("Enter list elements: ").split()))

# (a) Delete first occurrence of 32
listA.remove(32)

# (b) Delete all occurrences of 40
while 40 in listA:
    listA.remove(40)

# (c) Delete value at index = 5
del listA[5]

# (d) Modify value at index = 4 as 300 (originally 9, changed to fix IndexError)
listA[4] = 300

# (e) Add 200 at index = 2
listA.insert(2, 200)

# (f) Final output
print("Final list:", listA)
print("Length:", len(listA))
print("Maximum value:", max(listA))
print("Sum:", sum(listA))
```

```python
#string
s = input("Enter a string: ")

# (a) Delete value at index 4
s = s[:4] + s[5:]
print("After deleting index 4:", s)

# (b) Print string 4 times in one line
print("Repeated 4 times:", (s + " ") * 4)

# (c) Print string 3 times in new lines
print(s)
print(s)
print(s)

# (d) Modify character at index 7 as 'A'
s = s[:7] + 'A' + s[8:]
print("After modification:", s)

# (e) Print string from index 4 to 15
print("Substring (4 to 15):", s[4:16])

# (f) Final string and length
print("Final string:", s)
print("Length:", len(s))
```

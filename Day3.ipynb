```python

D = {
    "John": [25, 32, 43],
    "Peter": [87, 55, 96],
    "Ram": [58, 55, 43],
    "Meena": [63, 79, 85]
}


name = input("Enter name: ")

if name in D:
    D[name] = [88, 77, 99]
    print(D)
else:
    print("Name not found")
```

```python
age = int(input("Enter age: "))
gender = input("Enter gender (male/female): ").lower()

# Base ticket price
if age < 5:
    price = 0
elif 5 <= age <= 12:
    price = 10
elif age > 60:
    price = 0
else:
    price = 50

# Apply 50% discount for females
if gender == "female":
    price = price / 2

print("Ticket price is: Rs.", price)
```
